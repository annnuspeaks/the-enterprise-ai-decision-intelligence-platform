# Feature Scaling Strategy

This notebook identifies numerical features available for Machine Learning and recommends an appropriate scaling strategy.

Actual scaling is intentionally deferred to the individual Machine Learning pipelines to ensure reproducibility and avoid unnecessary preprocessing.

---

## Import Libraries
Import the required libraries for loading the engineered feature dataset and performing feature inspection.


```python
import pandas as pd
from pathlib import Path
```

## Load Engineered Feature Dataset
Load the engineered customer feature dataset generated during the Feature Creation phase.


```python
PROCESSED_DATA_DIR = Path("../data/processed")

customer_features = pd.read_csv(
    PROCESSED_DATA_DIR / "customer_features.csv"
)
```

## Identify Numerical Features
Extract all numerical features that may require scaling before model training.

```python
numerical_features = customer_features.select_dtypes(
    include="number"
).columns.tolist()

pd.DataFrame(
    {
        "Numerical Features": numerical_features
    }
)
```

## Recommend Scaling Strategy
Determine the recommended scaling method for each numerical feature.

```python
identifier_columns = [
    "customer_id",
    "customer_unique_id"
]

scaling_strategy = pd.DataFrame({
    "Feature": numerical_features,
    "Scaling Required": [
        "No" if feature in identifier_columns else "Yes"
        for feature in numerical_features
    ],
    "Recommended Scaler": [
        "None" if feature in identifier_columns else "StandardScaler / MinMaxScaler (Model Dependent)"
        for feature in numerical_features
    ]
})

scaling_strategy
```
