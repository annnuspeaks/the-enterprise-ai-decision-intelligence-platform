# Duplicate Detection

This notebook analyzes duplicate records across all tables of the Olist dataset.

The objective is to identify duplicate entries, measure their frequency, and determine whether data cleaning is required before feature engineering.

```python
import pandas as pd
from pathlib import Path
```

## Define Dataset Location

Specify the location of the raw Olist dataset files.

```python
DATA_DIR = Path("../data/raw/olist")
```

## Load All Dataset Tables

Load every relational dataset into memory for duplicate analysis.

```python
datasets = {
    "customers": pd.read_csv(DATA_DIR / "olist_customers_dataset.csv"),
    "orders": pd.read_csv(DATA_DIR / "olist_orders_dataset.csv"),
    "order_items": pd.read_csv(DATA_DIR / "olist_order_items_dataset.csv"),
    "payments": pd.read_csv(DATA_DIR / "olist_order_payments_dataset.csv"),
    "products": pd.read_csv(DATA_DIR / "olist_products_dataset.csv"),
    "reviews": pd.read_csv(DATA_DIR / "olist_order_reviews_dataset.csv"),
    "sellers": pd.read_csv(DATA_DIR / "olist_sellers_dataset.csv"),
    "geolocation": pd.read_csv(DATA_DIR / "olist_geolocation_dataset.csv"),
    "category_translation": pd.read_csv(
        DATA_DIR / "product_category_name_translation.csv"
    ),
}
```

## Detect Duplicate Records

Calculate the total number of duplicate rows and their percentage in each dataset.

```python
duplicate_summary = []

for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()

    duplicate_summary.append({
        "Dataset": name,
        "Rows": len(df),
        "Duplicate Rows": duplicate_count,
        "Duplicate (%)": round((duplicate_count / len(df)) * 100, 2)
    })

duplicate_summary_df = pd.DataFrame(duplicate_summary)

duplicate_summary_df
```

## Inspect Duplicate Records

For datasets containing duplicate rows, display a sample of duplicate records for manual inspection.

```python
for name, df in datasets.items():

    duplicates = df[df.duplicated()]

    if duplicates.empty:
        continue

    print(f"\n{name.upper()}")

    display(duplicates.head())
```

- Whether duplicates appear to be legitimate or accidental.
