# Data Cleaning Pipeline

This notebook performs the initial data cleaning process for the Olist Brazilian E-Commerce dataset.

The objective is to prepare high-quality datasets by removing inconsistencies while preserving business information for downstream analytics and Machine Learning tasks.

```python
import pandas as pd
from pathlib import Path
```

## Define Input and Output Directories

Specify the location of the raw datasets and the destination where cleaned datasets will be stored.

```python
RAW_DATA_DIR = Path("../data/raw/olist")
INTERIM_DATA_DIR = Path("../data/interim/olist")

INTERIM_DATA_DIR.mkdir(parents=True, exist_ok=True)
```

## Load All Dataset Tables

Load every relational dataset into memory before beginning the cleaning process.

```python
datasets = {
    "customers": pd.read_csv(RAW_DATA_DIR / "olist_customers_dataset.csv"),
    "orders": pd.read_csv(RAW_DATA_DIR / "olist_orders_dataset.csv"),
    "order_items": pd.read_csv(RAW_DATA_DIR / "olist_order_items_dataset.csv"),
    "payments": pd.read_csv(RAW_DATA_DIR / "olist_order_payments_dataset.csv"),
    "products": pd.read_csv(RAW_DATA_DIR / "olist_products_dataset.csv"),
    "reviews": pd.read_csv(RAW_DATA_DIR / "olist_order_reviews_dataset.csv"),
    "sellers": pd.read_csv(RAW_DATA_DIR / "olist_sellers_dataset.csv"),
    "geolocation": pd.read_csv(RAW_DATA_DIR / "olist_geolocation_dataset.csv"),
    "category_translation": pd.read_csv(
        RAW_DATA_DIR / "product_category_name_translation.csv"
    ),
}
```

## Create Working Copies

To preserve the original datasets, create independent working copies that will be modified during the cleaning process.

```python
cleaned_datasets = {
    name: df.copy(deep=True)
    for name, df in datasets.items()
}
```

## Remove Duplicate Records

Duplicate rows are removed from every dataset while preserving the first occurrence of each unique record.

```python
duplicate_report = []

for name, df in cleaned_datasets.items():

    before = len(df)

    df.drop_duplicates(inplace=True)

    after = len(df)

    duplicate_report.append({
        "Dataset": name,
        "Removed": before - after,
        "Remaining Rows": after
    })

duplicate_report = pd.DataFrame(duplicate_report)

duplicate_report
```

## Save Cleaned Datasets

Store the cleaned datasets inside the interim data directory for use in the remaining phases of the project.

```python
file_mapping = {
    "customers": "olist_customers_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "products": "olist_products_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "category_translation": "product_category_name_translation.csv",
}

for name, df in cleaned_datasets.items():
    df.to_csv(
        INTERIM_DATA_DIR / file_mapping[name],
        index=False
    )
```

## Conclusion

In this notebook:

- Raw datasets were preserved without modification.
- Independent working copies were created.
- Duplicate records were removed.
- Cleaned datasets were stored in the interim data directory.

These datasets will be used throughout the remaining data engineering pipeline.
