# Schema Validation

This notebook validates the structural integrity of the Olist Brazilian E-Commerce dataset before any preprocessing or feature engineering begins.

The validation includes loading all datasets, inspecting their dimensions, verifying column names, checking inferred data types, and ensuring that no duplicate columns exist.

```python
import pandas as pd
from pathlib import Path
```

## Define Dataset Location

The raw Olist datasets are stored under the project's `data/raw/olist/` directory. The following path object will be used to access all CSV files consistently throughout this notebook.

```python
DATA_DIR = Path("../data/raw/olist")
```

## Load All Dataset Tables

The Olist dataset is composed of multiple relational tables. Each table is loaded into a dictionary of pandas DataFrames for convenient access during validation and future preprocessing.

```python
datasets = {
    "customers": pd.read_csv(DATA_DIR / "olist_customers_dataset.csv"),
    "orders": pd.read_csv(DATA_DIR / "olist_orders_dataset.csv"),
    "order_items": pd.read_csv(DATA_DIR / "olist_order_items_dataset.csv"),
    "payments": pd.read_csv(DATA_DIR / "olist_order_payments_dataset.csv"),
    "products": pd.read_csv(DATA_DIR / "olist_products_dataset.csv"),
    "reviews": pd.read_csv(DATA_DIR / "olist_order_reviews_dataset.csv"),
    "sellers": pd.read_csv(DATA_DIR / "olist_sellers_dataset.csv"),
    "geolocation": pd.read_csv(DATA_DIR / "olist_geolocation_dataset.csv"),
    "category_translation": pd.read_csv(
        DATA_DIR / "product_category_name_translation.csv"
    ),
}

print(f"Loaded {len(datasets)} datasets successfully.")
```

## Verify Dataset Dimensions

The first validation step is to inspect the shape of every dataset. This helps confirm that all files have been loaded correctly and provides a quick overview of the number of records and attributes available in each table.

```python
for name, df in datasets.items():
    print(f"{name:<22} {df.shape}")
```

## Inspect Column Names

The schema of each dataset is examined by listing all column names. This ensures that the datasets match the expected structure and helps identify any missing or unexpected attributes.

```python
for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print("-" * 50)
    print(df.columns.tolist())
```

## Inspect Data Types

Pandas automatically infers data types while reading CSV files. Reviewing these inferred types helps identify columns that may require conversion during preprocessing, such as dates stored as text or numeric values interpreted incorrectly.

```python
for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print("-" * 50)
    print(df.dtypes)
```

## Check for Duplicate Columns

Duplicate column names can introduce ambiguity during joins, transformations, and feature engineering. This validation ensures that every dataset contains uniquely named columns.

```python
for name, df in datasets.items():
    duplicate_columns = df.columns[df.columns.duplicated()].tolist()

    print(f"{name:<22} Duplicate Columns : {duplicate_columns}")
```

## Generate Dataset Summary

Finally, a consolidated summary of all datasets is generated, displaying the number of rows and columns in each table. This serves as a quick reference for understanding the overall structure of the Olist dataset.

```python
summary = pd.DataFrame({
    "Rows": [df.shape[0] for df in datasets.values()],
    "Columns": [df.shape[1] for df in datasets.values()]
}, index=datasets.keys())

summary
```
