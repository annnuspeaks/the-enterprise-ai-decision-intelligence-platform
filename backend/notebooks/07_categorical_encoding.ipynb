# Categorical Encoding Strategy

This notebook identifies categorical variables present in the engineered dataset and defines the encoding strategy that will be applied during model development.

No permanent encoding is performed in this notebook because each Machine Learning pipeline will have its own preprocessing workflow.

## Import Required Libraries

```python
import pandas as pd
from pathlib import Path
```

## Define Dataset Location

Define the directory that contains the engineered customer feature dataset.

Using `pathlib` keeps file paths platform-independent.

```python
PROCESSED_DATA_DIR = Path("../data/processed")
```

## Load Engineered Customer Features

Load the engineered customer feature dataset that was generated during the previous Feature Engineering phase.

This dataset will be inspected to determine which columns require categorical encoding.

```python
customer_features = pd.read_csv(
    PROCESSED_DATA_DIR / "customer_features.csv"
)

customer_features.head()
```

## Inspect Dataset Information

Review the structure of the dataset including:

- Number of rows
- Number of columns
- Column names
- Data types
- Missing values

This helps identify categorical variables.

```python
customer_features.info()
```

## Identify Categorical Features

Extract every column whose datatype is `object`.

These columns are potential candidates for categorical encoding.

```python
categorical_columns = customer_features.select_dtypes(
    include="object"
).columns.tolist()

pd.DataFrame(
    {
        "Categorical Feature": categorical_columns
    }
)
```

## Classify Categorical Features

Determine how each categorical feature should be handled during Machine Learning.

Different feature types require different preprocessing strategies.

```python
strategy = []

for column in categorical_columns:

    if column.endswith("_id"):
        strategy.append("Identifier (Exclude from ML Features)")

    else:
        strategy.append("One-Hot Encoding")

encoding_strategy = pd.DataFrame(
    {
        "Feature": categorical_columns,
        "Recommended Strategy": strategy
    }
)

encoding_strategy
```

## Summary Statistics

Count how many categorical features belong to each encoding strategy.

This provides a quick overview of preprocessing requirements.

```python
encoding_strategy[
    "Recommended Strategy"
].value_counts().to_frame(
    name="Feature Count"
)
```

## Save Encoding Strategy

Save the encoding strategy so it can be referenced by future Machine Learning pipelines.

Only the strategy is saved.

The original dataset remains unchanged.

```python
encoding_strategy.to_csv(
    PROCESSED_DATA_DIR / "encoding_strategy.csv",
    index=False
)

print("Encoding strategy saved successfully.")
```

## Key observations:

- Identifier columns will not be used as Machine Learning features.
- Nominal categorical variables are recommended for One-Hot Encoding.
- No encoding has been applied in this notebook.
- Actual preprocessing will be implemented independently inside each Machine Learning pipeline using tools such as `ColumnTransformer` and `OneHotEncoder`.
