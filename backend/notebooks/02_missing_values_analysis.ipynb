# Missing Values Analysis

This notebook examines missing values across every table in the Olist dataset.

The objective is to identify incomplete data, measure its severity, and determine which columns will require cleaning or imputation before feature engineering.

```python
import pandas as pd
from pathlib import Path
```

## Define Dataset Location

The raw Olist datasets are loaded from the project's `data/raw/olist/` directory.

```python
DATA_DIR = Path("../data/raw/olist")
```

## Load All Dataset Tables

Load every relational dataset into memory so that missing values can be analyzed individually.

```python
datasets = {
    "customers": pd.read_csv(DATA_DIR / "olist_customers_dataset.csv"),
    "orders": pd.read_csv(DATA_DIR / "olist_orders_dataset.csv"),
    "order_items": pd.read_csv(DATA_DIR / "olist_order_items_dataset.csv"),
    "payments": pd.read_csv(DATA_DIR / "olist_order_payments_dataset.csv"),
    "products": pd.read_csv(DATA_DIR / "olist_products_dataset.csv"),
    "reviews": pd.read_csv(DATA_DIR / "olist_order_reviews_dataset.csv"),
    "sellers": pd.read_csv(DATA_DIR / "olist_sellers_dataset.csv"),
    "geolocation": pd.read_csv(DATA_DIR / "olist_geolocation_dataset.csv"),
    "category_translation": pd.read_csv(
        DATA_DIR / "product_category_name_translation.csv"
    ),
}
```

## Calculate Missing Values

For every dataset, calculate both the total count and percentage of missing values for each column.

```python
for name, df in datasets.items():

    missing = df.isnull().sum()

    missing = missing[missing > 0]

    if missing.empty:
        print(f"\n{name.upper()}")
        print("No missing values found.")
        continue

    report = pd.DataFrame({
        "Missing Values": missing,
        "Percentage": (missing / len(df) * 100).round(2)
    }).sort_values("Percentage", ascending=False)

    print(f"\n{name.upper()}")
    display(report)
```

## Missing Values Summary

Generate a concise summary showing the total number of missing values present in each dataset.

```python
summary = []

for name, df in datasets.items():

    total_missing = df.isnull().sum().sum()

    summary.append({
        "Dataset": name,
        "Total Missing Values": total_missing
    })

summary_df = pd.DataFrame(summary)

summary_df
```

## Key Observations

Review the generated reports and identify:

- Datasets with the highest number of missing values.
- Columns that may require imputation.
- Columns that may require removal.
- Columns suitable for leaving unchanged.

These findings will directly guide the Data Cleaning phase.
